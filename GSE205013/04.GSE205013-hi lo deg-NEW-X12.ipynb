```python
import omicverse as ov
#print(f"omicverse version: {ov.__version__}")
import scanpy as sc
#print(f"scanpy version: {sc.__version__}")
ov.utils.ov_plot_set()

import pandas as pd
import numpy as np
from scipy.stats import ttest_ind
from scipy.sparse import issparse
from scipy import sparse  
```

```python
adata = sc.read_h5ad('Ductal_cells_subtype_duc2_X7.h5ad')
adata
```

```python
lypla1_expr = get_expression_values(adata, 'LYPLA1')
adata.obs['LYPLA1_expression'] = lypla1_expr
adata.obs['LYPLA1_group'] = 'middle'

high_group_size = int(len(adata) * 0.35)
low_group_size = int(len(adata) * 0.35)
high_indices = adata.obs['LYPLA1_expression'].nlargest(high_group_size).index
low_indices = adata.obs['LYPLA1_expression'].nsmallest(low_group_size).index
adata.obs.loc[high_indices, 'LYPLA1_group'] = 'high'
adata.obs.loc[low_indices, 'LYPLA1_group'] = 'low'
sc.tl.rank_genes_groups(adata, groupby='LYPLA1_group', reference='low', method='wilcoxon')

diffexp_results = sc.get.rank_genes_groups_df(adata, group='high', key='rank_genes_groups')
diffexp_results = diffexp_results[['names', 'logfoldchanges', 'pvals_adj']].dropna()
diffexp_results = diffexp_results.drop_duplicates(subset=['names'])
output_filename = "X12.GSEA/X12.LYPLA1_35_all_genes_with_logfc.csv"
diffexp_results.to_csv(output_filename, index=False)

print(f"Differential expression analysis complete. Results saved to {output_filename}.")
```
